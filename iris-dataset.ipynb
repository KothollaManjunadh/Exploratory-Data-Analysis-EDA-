```python
# SkLearn Imports
from sklearn.datasets import load_iris
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.ensemble import RandomForestClassifier

# Data Analytics Imports
import numpy as np
import pandas as pd

# Data Visualisation Imports
import plotly.express as px
import plotly.graph_objects as go
from plotly.offline import plot, iplot, init_notebook_mode, download_plotlyjs
```

## Loading the Dataset

```python
# Loading the dataset
iris = load_iris()
print(iris.keys())
```

```python
print("Features Used: ")
print(iris["feature_names"])
```

```python
print("Targets Used: ")
print(iris["target_names"])
```

```python
print("Description of the Dataset (similar to pd.DataFrame.describe()): ")
print(iris["DESCR"])
```

## Converting Arrays into a Pandas DataFrame

```python
df_iris = pd.DataFrame(iris["data"], columns=iris["feature_names"])
df_iris.head()
```

```python
df_species = pd.DataFrame(iris["target"], columns=["species"])
df_species.head()
```

```python
df_iris_full = pd.concat([df_iris, df_species], axis=1)
df_iris_full.head()
```

```python
df_iris_full.info()
```

## Creating a Training and Testing Set

```python
# Making a 10 fold split of the dataset by default
split = StratifiedShuffleSplit(n_splits=1, test_size=0.10, random_state=17)
for train_idx, test_idx in split.split(df_iris_full, df_iris_full["species"]):
    strat_train = df_iris_full.iloc[train_idx]
    strat_test = df_iris_full.iloc[test_idx]
```

```python
print(f"Shape of the Training Set: {strat_train.shape}")
strat_train.head()
```

```python
print(f"Shape of the Testing Set: {strat_test.shape}")
strat_test.head()
```

## Visualising and Gaining Insights

```python
targets = iris["target_names"]

fig = px.scatter_3d(df_iris_full, x="petal length (cm)", y="petal width (cm)", z="sepal length (cm)", color="species",
                   hover_name=targets[df_iris_full["species"]],
                   labels={
                       "petal length (cm)": "Petal Length (cm)",
                       "petal width (cm)": "Petal Width (cm)",
                       "sepal length (cm)": "Sepal Length (cm)"
                   })
fig.update_traces(marker_line_width=4, marker_size=10)
```

**`Important`**  
From the above graph we can infer the following:
- `Virginica`: Highest Petal Length, Highest Sepal Length, Highest Sepal Width
- `Sentosa`: Lowest Petal Length, Lowest Sepal Length, Lowest Sepal Width

```python
# Finding the correlation among the attributes
corr_matrix = df_iris_full.corr()
corr_matrix["species"].sort_values(ascending=False)
```

**`Inference`**  
- The results of the `Scatter Plot` are verified by the `Correlation Calculation`
- The `Petal Length` and `Petal Width` have very high `Positive Correlation` with the target attribute

## Visualising the Distribution of the Attributes

```python
fig = px.scatter_matrix(df_iris_full.drop("species", axis=1))
fig.update_layout(font=dict(size=8))
```

## Building the Model

```python
# Seperating the Training Set
X_train = strat_train.drop("species", axis=1)
y_train = strat_train["species"]

# Seperating the Test Set
X_test = strat_test.drop("species", axis=1)
y_test = strat_test["species"]
```

```python
# Initialising the Model
tree_classifier = RandomForestClassifier()

# Training the Model
tree_classifier.fit(X_train, y_train)
```

## Making Predictions

```python
predictions = tree_classifier.predict(X_test)
n_correct = sum(predictions == y_test)
print("Score of Predictions: ", n_correct / len(predictions))
```
